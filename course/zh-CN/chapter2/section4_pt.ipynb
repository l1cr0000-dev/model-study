```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")

sequence = "Using a Transformer network is simple"
tokens = tokenizer.tokenize(sequence)

print(tokens)
```

```python
ids = tokenizer.convert_tokens_to_ids(tokens)

print(ids)
```

```python
s = [['I’ve been waiting for a HuggingFace course my whole life.']
     ,['和“I hate this so much!']]
s1 = 'I’ve been waiting for a HuggingFace course my whole life.'
s2 = 'I hate this so much!'
```

```python
tokens1 = tokenizer.tokenize(s1)
print(tokens1)
ids1 = tokenizer.convert_tokens_to_ids(tokens1)
print(ids1)
```

```python
tokens2 = tokenizer.tokenize(s2)
print(tokens2)
ids2 = tokenizer.convert_tokens_to_ids(tokens2)
print(ids2)
```
